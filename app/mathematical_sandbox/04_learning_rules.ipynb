```python
# Hebbian Learning

# Initial weight
weight = 0.2

# Learning rate
learning_rate = 0.1

# Input and output neuron activation
input_value = 1
output_value = 1

print("Initial Weight:", weight)

# Hebbian weight update
weight_change = learning_rate * input_value * output_value

weight = weight + weight_change

print("Weight Change:", weight_change)
print("Updated Weight:", weight)
```

```python
# Modified Hebbian Learning (Oja's Rule)

weight = 0.2
learning_rate = 0.1

input_value = 1.0

print("Initial Weight:", weight)

for step in range(5):

    # Neuron output
    output_value = weight * input_value

    # Oja's learning rule
    weight_change = learning_rate * output_value * (
        input_value - output_value * weight
    )

    weight += weight_change

    print(
        f"Step {step + 1}: "
        f"Weight = {weight:.4f}"
    )

print("\nFinal Weight:", weight)
```

```python
import numpy as np

# Input vector
input_vector = np.array([0.8, 0.6])

# Weights of three neurons
weights = np.array([
    [0.2, 0.3],
    [0.7, 0.5],
    [0.1, 0.9]
])

learning_rate = 0.5

print("Input Vector:", input_vector)

# Calculate Euclidean distance
distances = np.linalg.norm(weights - input_vector, axis=1)

print("\nDistances:", distances)

# Find winning neuron
winner = np.argmin(distances)

print("\nWinning Neuron:", winner + 1)

# Update only the winning neuron's weights
weights[winner] += learning_rate * (
    input_vector - weights[winner]
)

print("\nUpdated Weights:")
print(weights)
```

```python
import numpy as np

# Training data
inputs = np.array([1.0, 0.5])

# Target output
target = 1.0

# Initial weights and bias
weights = np.array([0.2, 0.3])
bias = 0.1

learning_rate = 0.1

print("Initial Weights:", weights)

# Prediction
prediction = np.dot(inputs, weights) + bias

# Calculate error
error = target - prediction

print("Prediction:", prediction)
print("Error:", error)

# Update weights and bias
weights += learning_rate * error * inputs
bias += learning_rate * error

print("\nUpdated Weights:", weights)
print("Updated Bias:", bias)

# New prediction
new_prediction = np.dot(inputs, weights) + bias

print("\nNew Prediction:", new_prediction)
```

```python
import numpy as np

np.random.seed(42)

# Sigmoid activation
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# One training example with two visible units
visible_data = np.array([[1.0, 0.0]])

# Two visible units, one hidden unit
weights = np.array([
    [0.2],
    [0.3]
])

visible_bias = np.zeros((1, 2))
hidden_bias = np.zeros((1, 1))

learning_rate = 0.1

# Positive phase: calculate hidden probabilities
hidden_prob = sigmoid(
    visible_data @ weights + hidden_bias
)

# Sample hidden state
hidden_state = (
    np.random.random(hidden_prob.shape) < hidden_prob
).astype(float)

# Negative phase: reconstruct visible units
visible_prob = sigmoid(
    hidden_state @ weights.T + visible_bias
)

visible_state = (
    np.random.random(visible_prob.shape) < visible_prob
).astype(float)

# Calculate hidden probabilities again
hidden_prob_negative = sigmoid(
    visible_state @ weights + hidden_bias
)

# Contrastive Divergence weight update
positive_association = visible_data.T @ hidden_prob
negative_association = visible_state.T @ hidden_prob_negative

weights += learning_rate * (
    positive_association - negative_association
)

# Update biases
visible_bias += learning_rate * (
    visible_data - visible_state
)

hidden_bias += learning_rate * (
    hidden_prob - hidden_prob_negative
)

print("Original Input:", visible_data)
print("\nHidden Probabilities:", hidden_prob)
print("Sampled Hidden State:", hidden_state)
print("\nReconstructed Input:", visible_state)
print("\nUpdated Weights:")
print(weights)

print("\nUpdated Visible Bias:", visible_bias)
print("Updated Hidden Bias:", hidden_bias)
```
