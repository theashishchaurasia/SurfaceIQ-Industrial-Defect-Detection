```python
# McCulloch-Pitts Neuron

def mp_neuron(inputs, weights, threshold):

    weighted_sum = sum(
        x * w for x, w in zip(inputs, weights)
    )

    if weighted_sum >= threshold:
        return 1
    else:
        return 0


# AND Gate Example
inputs = [1, 1]
weights = [1, 1]
threshold = 2

output = mp_neuron(inputs, weights, threshold)

print("Inputs:", inputs)
print("Weights:", weights)
print("Threshold:", threshold)
print("Neuron Output:", output)
```

```python
import numpy as np

# Input layer: 2 neurons
inputs = np.array([0.5, 0.8])

# Weights between input and hidden layer
W1 = np.array([
    [0.2, 0.4],
    [0.3, 0.5]
])

# Hidden layer bias
b1 = np.array([0.1, 0.1])

# Weights between hidden and output layer
W2 = np.array([
    [0.6],
    [0.7]
])

# Output layer bias
b2 = np.array([0.2])

# Sigmoid activation
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Hidden layer calculation
hidden = sigmoid(np.dot(inputs, W1) + b1)

# Output layer calculation
output = sigmoid(np.dot(hidden, W2) + b2)

print("Input:", inputs)
print("Hidden Layer Output:", hidden)
print("Final Network Output:", output)
```

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Input
X = np.array([0.5, 0.8])

# Weights and biases
W1 = np.array([
    [0.2, 0.4],
    [0.3, 0.5]
])
b1 = np.array([0.1, 0.1])

W2 = np.array([
    [0.6],
    [0.7]
])
b2 = np.array([0.2])

# Step 1: Weighted sum at hidden layer
Z1 = np.dot(X, W1) + b1

# Step 2: Hidden layer activation
A1 = sigmoid(Z1)

# Step 3: Weighted sum at output layer
Z2 = np.dot(A1, W2) + b2

# Step 4: Final prediction
A2 = sigmoid(Z2)

print("Input:", X)
print("Hidden Layer Weighted Sum:", Z1)
print("Hidden Layer Activation:", A1)
print("Output Layer Weighted Sum:", Z2)
print("Final Prediction:", A2)
```

```python
import numpy as np

# Training data
X = np.array([[1.0]])
Y = np.array([[1.0]])

# Initial weight and bias
W = np.array([[0.5]])
b = np.array([[0.0]])

learning_rate = 0.1

# Sigmoid activation
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Step 1: Forward propagation
Z = np.dot(X, W) + b
prediction = sigmoid(Z)

# Step 2: Calculate Mean Squared Error
loss = np.mean((prediction - Y) ** 2)

# Step 3: Backpropagation
dL_dA = 2 * (prediction - Y)
dA_dZ = prediction * (1 - prediction)

dL_dZ = dL_dA * dA_dZ

dW = np.dot(X.T, dL_dZ)
db = np.sum(dL_dZ)

# Step 4: Update weights and bias
W_new = W - learning_rate * dW
b_new = b - learning_rate * db

print("Prediction:", prediction)
print("Loss:", loss)
print("Weight Gradient:", dW)
print("Bias Gradient:", db)
print("Updated Weight:", W_new)
print("Updated Bias:", b_new)
```
