```python
import numpy as np

# Creating two matrices (2D tensors)
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

# Addition
print("\nMatrix Addition:")
print(A + B)

# Element-wise multiplication
print("\nElement-wise Multiplication:")
print(A * B)

# Matrix multiplication
print("\nMatrix Multiplication:")
print(A @ B)
```

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

print("Original Matrix A:")
print(A)

# Perform SVD
U, S, VT = np.linalg.svd(A)

print("\nMatrix U:")
print(U)

print("\nSingular Values:")
print(S)

print("\nMatrix V Transpose:")
print(VT)

# Reconstruct original matrix
Sigma = np.zeros((3, 2))
np.fill_diagonal(Sigma, S)

reconstructed = U @ Sigma @ VT

print("\nReconstructed Matrix:")
print(np.round(reconstructed, 2))
```

```python
import numpy as np

# Define a simple mathematical function
def f(x):
    return x**2 + 3*x + 2

# Analytical derivative
def analytical_gradient(x):
    return 2*x + 3

# Numerical gradient using finite differences
def numerical_gradient(x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2*h)

x = 4.0

analytical = analytical_gradient(x)
numerical = numerical_gradient(x)

print("Input x:", x)
print("Analytical Gradient:", analytical)
print("Numerical Gradient:", numerical)

print("\nGradient Difference:", abs(analytical - numerical))

if np.isclose(analytical, numerical, atol=1e-5):
    print("\nGradient Check Passed!")
else:
    print("\nGradient Check Failed!")
```
